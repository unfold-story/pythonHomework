```python
#3.1
from scipy.linalg import solve
A = np.array([
    [3, 1],
    [1, 2]
])
b = np.array([9, 8])
solution = solve(A, b)
x, y = solution
print("x =", x)
print("y =", y)
verification = A @ solution
print("代入验证结果：", verification)
print("是否满足方程：", np.allclose(verification, b))
```

```python
#3.2
from scipy.optimize import root

def equation(x):
    return x[0] ** 2 - 2

positive_result = root(equation, [1.0])
negative_result = root(equation, [-1.0])
print("正根求解成功：", positive_result.success)
print("正根：", positive_result.x[0])
print("正根函数值：", equation(positive_result.x))
print()
print("负根求解成功：", negative_result.success)
print("负根：", negative_result.x[0])
print("负根函数值：", equation(negative_result.x))
print(
    "正根是否满足方程：",
    np.isclose(equation(positive_result.x), 0)
)
print(
    "负根是否满足方程：",
    np.isclose(equation(negative_result.x), 0)
)
```

```python
#3.3
from scipy.optimize import minimize
def objective(x):
    return x[0] ** 2 + x[0] + 2
result = minimize(objective, x0=[0.0])
print("优化是否成功：", result.success)
print("极小点：", result.x[0])
print("最小函数值：", result.fun)
analytical_x = -0.5
analytical_f = objective([analytical_x])
print("解析极小点：", analytical_x)
print("解析最小值：", analytical_f)
print(
    "极小点是否接近：",
    np.isclose(result.x[0], analytical_x)
)
```

```python
#3.4
from scipy.stats import ttest_ind
rng = np.random.default_rng(42)
sample_a = rng.normal(
    loc=0,
    scale=1,
    size=100
)
sample_b = rng.normal(
    loc=1,
    scale=1,
    size=100
)
t_statistic, p_value = ttest_ind(
    sample_a,
    sample_b
)
print("第一组样本均值：", np.mean(sample_a))
print("第二组样本均值：", np.mean(sample_b))
print("t统计量：", t_statistic)
print("p值：", p_value)
alpha = 0.05
if p_value < alpha:
    print("p < 0.05，拒绝原假设")
    print("两组总体均值存在统计学显著差异")
else:
    print("p >= 0.05，未能拒绝原假设")
    print("没有足够证据认为两组总体均值存在差异")
```

```python
#4.1
import pandas as pd
records = {
    "Name": ["Alice", "Bob", "Carol", "David", "Eva", "Frank"],
    "Group": ["A", "A", "B", "B", "A", "B"],
    "Math": [88, 72, 91, 65, 79, 84],
    "English": [90, 68, 85, 73, 92, 77],
    "Python": [95, 80, 89, 70, 86, 82],
}

df = pd.DataFrame(records)
df = df.set_index("Name")
subjects = ["Math", "English", "Python"]
df["Total"] = df[subjects].sum(axis=1)
df["Average"] = df[subjects].mean(axis=1)
print("行索引：", df.index)
print("\n列名：", df.columns)
print("\n前3行：")
print(df.head(3))
print("\n基本统计信息：")
print(df.describe())
print("\n实验2模拟成绩表格：")
print(scores_table.head())
```

```python
#4.2
carol_loc = df.loc["Carol"]
carol_iloc = df.iloc[2]
print("loc读取Carol：")
print(carol_loc)
print("\niloc读取Carol：")
print(carol_iloc)
high_score_students = df[df["Average"] >= 80]
print("\n平均分不低于80的学生：")
print(high_score_students)
```

```python
#4.3
subject_means = df[subjects].mean(axis=0)
student_means = df[subjects].mean(axis=1)
print("三科各自的平均分：")
print(subject_means)
print("\n每位学生的三科平均分：")
print(student_means)
```

```python
#4.4
df["Highest"] = df[subjects].apply(
    lambda row: row.max(),
    axis=1
)
print("每位学生的最高分：")
print(df["Highest"])
statistics = df[subjects].agg(
    ["min", "mean", "max"]
)
print("\n各科统计结果：")
print(statistics)
df["Math_Rank"] = df["Math"].rank(
    ascending=False,
    method="min"
)
print("\nMath成绩排名：")
print(df[["Math", "Math_Rank"]].sort_values("Math_Rank"))
```

```python
#4.5
grouped = df.groupby("Group")
group_counts = grouped.size()
print("各组人数：")
print(group_counts)
group_means = grouped[subjects].mean()
print("\n各组三科平均分：")
print(group_means)
df["Group_Python_Mean"] = grouped["Python"].transform("mean")
print("\n增加所在组Python平均分：")
print(df[["Group", "Python", "Group_Python_Mean"]])
```

```python
#4.6
filtered_df = df.groupby("Group").filter(
    lambda group: group["Python"].mean() >= 80
)
retained_groups = filtered_df["Group"].unique()
print("保留的组：", retained_groups)
print("\n保留的学生记录：")
print(filtered_df)
```

```python
#4.7
score_data = df[subjects]
cov_matrix = score_data.cov()
corr_matrix = score_data.corr(method="pearson")
print("协方差矩阵：")
print(cov_matrix)
print("\n皮尔森相关系数矩阵：")
print(corr_matrix)

cov_value = cov_matrix.loc["Math", "Python"]
corr_value = corr_matrix.loc["Math", "Python"]
print("\nMath与Python的协方差：", cov_value)
print("Math与Python的相关系数：", corr_value)

if cov_value > 0:
    print("协方差为正：两科成绩总体呈同向变化")
elif cov_value < 0:
    print("协方差为负：两科成绩总体呈反向变化")
else:
    print("协方差为0：没有观察到线性协同变化")

print("相关系数取值范围为[-1, 1]")
print("绝对值越接近1，样本中的线性相关程度越强")
print("绝对值越接近0，线性相关程度越弱")
```

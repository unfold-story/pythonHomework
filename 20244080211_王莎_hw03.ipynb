```python
#1.1
import numpy as np

def wigner_density(x):
    if abs(x) > 2:
        return 0.0
    return float(np.sqrt(4 - x ** 2) / (2 * np.pi))

print(wigner_density(0))
print(wigner_density(1))
print(wigner_density(2))
print(wigner_density(3))
print(wigner_density(np.float64(-1.5)))
```

```python
#1.2
import numpy as np

rng = np.random.default_rng(42)

def generate_wigner(n):
    if isinstance(n, (bool, np.bool_)) or not isinstance(n, (int, np.integer)):
        raise TypeError("n必须是正整数")
    if n <= 0:
        raise ValueError("n必须大于0")

    upper = np.triu(
        rng.normal(0, 1 / np.sqrt(n), size=(n, n))
    )

    matrix = upper + np.triu(upper, 1).T
    return matrix


Z = generate_wigner(4)

print("Wigner矩阵：")
print(Z)
print("矩阵形状：", Z.shape)
print("是否对称：", np.allclose(Z, Z.T))

try:
    generate_wigner(-2)
except ValueError as e:
    print("异常测试：", e)
```

```python
#1.3
import numpy as np

def get_spectrum(matrix):
    eigenvalues = np.linalg.eigvalsh(matrix)
    return np.sort(np.asarray(eigenvalues))

matrix = np.array([
    [2, 1],
    [1, 2]
])
spectrum = get_spectrum(matrix)

print("原矩阵：")
print(matrix)
print("特征值：", spectrum)
print("是否按非递减顺序排列：", np.all(np.diff(spectrum) >= 0))
```

```python
#1.4
import numpy as np
import matplotlib.pyplot as plt

sizes = [100, 200, 500, 1000]
rng = np.random.default_rng(42)

x = np.linspace(-2.2, 2.2, 500)
density = np.sqrt(np.maximum(4 - x ** 2, 0)) / (2 * np.pi)

fig, axes = plt.subplots(4, 1, figsize=(9, 17))

for ax, n in zip(axes, sizes):
    matrix = generate_wigner(n)
    eigenvalues = get_spectrum(matrix)
    ax.hist(
        eigenvalues,
        bins=35,
        density=True,
        color="blue",
        alpha=0.65,
        edgecolor="white",
        label="Eigenvalue Distribution"
    )
    ax.plot(
        x,
        density,
        color="red",
        linewidth=2,
        label="Wigner Semicircle Density"
    )
    ax.set_title(f"Wigner Semicircle Law (n={n})")
    ax.set_xlabel("Eigenvalue")
    ax.set_ylabel("Probability Density")
    ax.set_xlim(-2.2, 2.2)
    ax.legend()
    ax.grid(alpha=0.2)

plt.tight_layout()
plt.show()
```

```python
#2.1
import numpy as np

def is_valid_board(board):
    if not isinstance(board, np.ndarray):
        return False

    if board.ndim != 2:
        return False

    if board.shape[0] == 0 or board.shape[1] == 0:
        return False

    return bool(np.all((board == 0) | (board == 1)))

board1 = np.array([
    [0, 1, 0],
    [1, 1, 0],
    [0, 0, 1]
])

board2 = np.array([
    [0, 1, 2],
    [1, 0, 0]
])

board3 = np.array([0, 1, 0])

print(is_valid_board(board1))
print(is_valid_board(board2))
print(is_valid_board(board3))
print(is_valid_board([[0, 1], [1, 0]]))
```

```python
#2.2
import numpy as np

def gol_step(board):
    if not is_valid_board(board):
        raise ValueError("无效的生命游戏棋盘")

    neighbors = np.zeros(board.shape, dtype=int)

    for dx in [-1, 0, 1]:
        for dy in [-1, 0, 1]:
            if dx == 0 and dy == 0:
                continue

            neighbors += np.roll(
                np.roll(board, dx, axis=0),
                dy,
                axis=1
            )

    next_board = (
        ((board == 1) & ((neighbors == 2) | (neighbors == 3)))
        | ((board == 0) & (neighbors == 3))
    ).astype(int)

    return next_board

board = np.zeros((5, 5), dtype=int)
board[2, 1:4] = 1

print("原棋盘：")
print(board)
next_board = gol_step(board)
print("\n下一代棋盘：")
print(next_board)
print("\n原棋盘是否保持不变：")
print(board)
try:
    gol_step(np.array([[0, 2], [1, 0]]))
except ValueError as e:
    print("异常测试：", e)
```

```python
#2.3
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap

def draw_gol_board(board):
    if not is_valid_board(board):
        raise ValueError("无效的生命游戏棋盘")
    cmap = ListedColormap(["white", "black"])
    fig, ax = plt.subplots(figsize=(6, 6))
    ax.imshow(
        board,
        cmap=cmap,
        vmin=0,
        vmax=1,
        interpolation="nearest"
    )
    ax.set_xticks([])
    ax.set_yticks([])
    ax.axis("off")

    plt.show()

board = np.zeros((10, 10), dtype=int)
board[4, 3:6] = 1
draw_gol_board(board)
next_board = gol_step(board)
draw_gol_board(next_board)
```

```python
#1.1
import re

def is_alphanumeric(s):
    return re.fullmatch(r"[a-zA-Z0-9]+", s) is not None

print(is_alphanumeric("Hello123"))
print(is_alphanumeric("Python"))
print(is_alphanumeric("123456"))
print(is_alphanumeric("Hello World"))
print(is_alphanumeric("abc_123"))
print(is_alphanumeric("你好123"))
print(is_alphanumeric(""))
```

```python
#1.2
import re

def match_zero_or_more_b(s):
    return re.fullmatch(r"ab*", s) is not None

print(match_zero_or_more_b("a"))
print(match_zero_or_more_b("ab"))
print(match_zero_or_more_b("abb"))
print(match_zero_or_more_b("abbbb"))
print(match_zero_or_more_b("ac"))
print(match_zero_or_more_b("b"))
```

```python
#1.3
import re

def match_one_or_more_b(s):
    return re.fullmatch(r"ab+", s) is not None

print(match_one_or_more_b("a"))
print(match_one_or_more_b("ab"))
print(match_one_or_more_b("abb"))
print(match_one_or_more_b("abbbb"))
print(match_one_or_more_b("abc"))
print(match_one_or_more_b("b"))
```

```python
#1.4
import re

def match_a_to_b(s):
    return re.fullmatch(r"a[\s\S]*b", s) is not None

print(match_a_to_b("ab"))
print(match_a_to_b("acb"))
print(match_a_to_b("a123b"))
print(match_a_to_b("a hello b"))
print(match_a_to_b("a\nb"))
print(match_a_to_b("abc"))
print(match_a_to_b("ba"))
```

```python
#1.5
import re

def remove_ip_leading_zeros(ip):
    parts = ip.split(".")
    result = []

    for part in parts:
        cleaned = re.sub(r"^0+(?=\d)", "", part)
        result.append(cleaned)

    return ".".join(result)

print(remove_ip_leading_zeros("192.168.001.010"))
print(remove_ip_leading_zeros("001.002.003.004"))
print(remove_ip_leading_zeros("000.000.000.000"))
print(remove_ip_leading_zeros("010.020.030.040"))
print(remove_ip_leading_zeros("127.0.0.1"))
```

```python
#1.6
import re

def swap_space_underscore(s):
    return re.sub(
        r"[ _]",
        lambda match: "_" if match.group() == " " else " ",
        s
    )

print(swap_space_underscore("Hello World"))
print(swap_space_underscore("Hello_World"))
print(swap_space_underscore("Python_Data Analysis"))
print(swap_space_underscore("a_b c_d"))
```

```python
#1.7
import re

def convert_date(s):
    return re.sub(
        r"\b(\d{4})-(\d{2})-(\d{2})\b",
        r"\3-\2-\1",
        s
    )

print(convert_date("2024-05-01"))
print(convert_date("2026-11-25"))
print(convert_date("2000-01-15"))
print(convert_date("Today is 2025-12-31"))
```

```python
#1.8
import re

def find_words_starting_ae(s):
    return re.findall(
        r"\b[ae][a-z]*\b",
        s,
        flags=re.IGNORECASE
    )

text = "Apple and elephant are eating an orange."

print(find_words_starting_ae(text))
print(find_words_starting_ae("An egg is on a table"))
print(find_words_starting_ae("Python Java Kotlin"))
print(find_words_starting_ae("Excellent algorithms are amazing"))
```

```python
#1.9
import re

def abbreviate_road(s):
    return re.sub(r"\bRoad\b", "Rd", s)

print(abbreviate_road("Oxford Road"))
print(abbreviate_road("Main Road, London"))
print(abbreviate_road("Road Road"))
print(abbreviate_road("Roadside"))
print(abbreviate_road("Broadway Road"))
```

```python
#1.10
import re

def find_long_words(s):
    return re.findall(r"\b[a-zA-Z]{4,}\b", s)

text = "The quick brown fox jumps over the lazy dog"

print(find_long_words(text))
print(find_long_words("I love Python programming"))
print(find_long_words("a an cat bird elephant"))
print(find_long_words("Hi to me"))
```

```python
#1.11
import re

def remove_non_alphanumeric(s):
    return re.sub(r"[^a-zA-Z0-9]", "", s)

print(remove_non_alphanumeric("Hello, World!"))
print(remove_non_alphanumeric("Python_3.11"))
print(remove_non_alphanumeric("abc@123#xyz"))
print(remove_non_alphanumeric("A B C 123"))
print(remove_non_alphanumeric("你好Python123"))
```

```python
#1.12
import re

def remove_parentheses(s):
    result = re.sub(r"\([^()]*\)", "", s)
    return result.strip()

examples = [
    "example (.com)",
    "w3resource",
    "github (.com)",
    "stackoverflow (.com)"
]

for text in examples:
    print(remove_parentheses(text))

print(remove_parentheses("Python (version 3.11)"))
print(remove_parentheses("Hello (abc) World (xyz)"))
```

```python
#2.1
import requests
import re
from datetime import date
from bs4 import BeautifulSoup
from bs4.element import Tag

url = "https://en.wikipedia.org/wiki/Madison,_Wisconsin"

headers = {
    "User-Agent": "PythonHomework/1.0 (educational project; requests)"
}

response = requests.get(
    url,
    headers=headers,
    timeout=30
)
response.raise_for_status()

soup = BeautifulSoup(response.text, "html.parser")

climate_table = None

for table in soup.find_all("table"):
    text = re.sub(r"\s+", " ", table.get_text(" ", strip=True))

    if re.search(r"Climate data for", text, re.IGNORECASE):
        climate_table = table
        break

print("访问URL：", url)
print("访问日期：", date.today().isoformat())
print("HTTP状态码：", response.status_code)

if climate_table is not None:
    children = [
        child
        for child in climate_table.children
        if isinstance(child, Tag)
    ]

    print("找到气候数据表格")
    print("表格标签：", climate_table.name)
    print("直接子节点数量：", len(children))
    print("直接子节点名称：", [child.name for child in children])

    print("\n表格前500个字符：")
    print(climate_table.get_text(" ", strip=True)[:500])
else:
    print("页面读取成功，但未找到气候数据表格")
```

```python
#2.2
import re
import requests
from bs4 import BeautifulSoup
from bs4.element import Tag

def retrieve_climate_table(s):
    response = requests.get(
        s,
        headers={
            "User-Agent": "PythonHomework/1.0 (educational project; requests)"
        },
        timeout=30
    )

    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")

    for table in soup.find_all("table"):
        captions = table.find_all("caption")
        first_rows = table.find_all("tr", limit=4)

        candidates = []

        for caption in captions:
            candidates.append(caption.get_text(" ", strip=True))

        for row in first_rows:
            candidates.append(row.get_text(" ", strip=True))

        candidates.append(
            table.get_text(" ", strip=True)[:500]
        )

        previous_heading = table.find_previous(
            ["h2", "h3", "h4"]
        )

        if previous_heading is not None:
            candidates.append(
                previous_heading.get_text(" ", strip=True)
            )

        normalized = [
            re.sub(r"\s+", " ", text).strip()
            for text in candidates
        ]

        if any(
            re.search(r"\bClimate data for\b", text, re.IGNORECASE)
            for text in normalized
        ):
            return table

    return None


urls = {
    "Madison": "https://en.wikipedia.org/wiki/Madison,_Wisconsin",
    "Boston": "https://en.wikipedia.org/wiki/Boston",
    "Los Angeles": "https://en.wikipedia.org/wiki/Climate_of_Los_Angeles"
}

for city, url in urls.items():
    print("\n城市：", city)
    print("URL：", url)

    try:
        table = retrieve_climate_table(url)

        if table is None:
            print("成功访问，但没有找到气候数据表格")
        else:
            print("返回类型：", type(table).__name__)
            print("表格标签：", table.name)
            print("表格前200个字符：")
            print(table.get_text(" ", strip=True)[:200])

    except requests.exceptions.RequestException as e:
        print("网络或HTTP错误：", e)
```

```python
#2.3
import re
import requests

def list_climate_table_row_names(s):
    table = retrieve_climate_table(s)

    if table is None:
        return None

    row_names = []

    for tr in table.find_all("tr"):
        cells = tr.find_all(
            ["th", "td"],
            recursive=False
        )

        if not cells:
            continue

        first_cell = cells[0]

        if first_cell.name != "th":
            continue

        if len(cells) < 2:
            continue

        name = re.sub(
            r"\s+",
            " ",
            first_cell.get_text(" ", strip=True)
        ).strip()

        if not name:
            continue

        lower_name = name.lower()

        if lower_name.startswith("climate data for"):
            continue

        if lower_name.startswith("month"):
            continue

        if lower_name.startswith("source"):
            continue

        if lower_name.startswith("see also"):
            continue

        if lower_name.startswith("notes"):
            continue

        if not tr.find_all("td", recursive=False):
            continue

        row_names.append(name)

    return row_names


test_urls = {
    "Madison": "https://en.wikipedia.org/wiki/Madison,_Wisconsin",
    "Boston": "https://en.wikipedia.org/wiki/Boston",
    "Los Angeles": "https://en.wikipedia.org/wiki/Climate_of_Los_Angeles"
}

for city, url in test_urls.items():
    print("\n城市：", city)

    try:
        names = list_climate_table_row_names(url)

        if names is None:
            print("没有找到气候数据表格")
        else:
            print("行名数量：", len(names))
            print("行名列表：")

            for name in names:
                print(name)

            print(
                "是否全部为字符串：",
                all(isinstance(name, str) for name in names)
            )

    except requests.exceptions.RequestException as e:
        print("网络或HTTP错误：", e)
```

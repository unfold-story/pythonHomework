```python
#1.1
def is_palindrome(s):
    cleaned = s.replace(" ", "")
    cleaned = cleaned.lower()
    reversed_s = cleaned[::-1]
    return cleaned == reversed_s

print(is_palindrome("T A C O cat"))
```

```python
#1.2
def is_abecedarian(s):
    cleaned = s.replace(" ", "").lower()
    for i in range(len(cleaned) - 1):
        if cleaned[i] > cleaned[i + 1]:
            return False
    return True
```

```python
#1.3
def remove_vowels(s):
    vowels = "aeiou"
    result = ""
    for char in s:
        if char.lower() not in vowels:
            result += char
    return result
```

```python
#2.1
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("输入参数必须是列表list")
    return lst[::-1]
```

```python
#2.2
def is_sorted(seq):
    for i in range(1, len(seq)):
        if seq[i - 1] > seq[i]:
            return False
    return True
```

```python
#3.1
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("向量必须是tuple类型")
    if type(s) not in (int, float):
        raise TypeError("标量必须是整数或浮点数")
    for value in t:
        if type(value) not in (int, float):
            raise TypeError("向量元素必须是整数或浮点数")
    result = tuple(value * s for value in t)
    return result
```

```python
#3.2
def vec_inner_product(t1, t2):
    if not isinstance(t1, tuple) or not isinstance(t2, tuple):
        raise TypeError("两个向量都必须是tuple类型")
    if len(t1) != len(t2):
        raise ValueError("两个向量的长度必须相同")
    for value in t1 + t2:
        if type(value) not in (int, float):
            raise TypeError("向量元素必须是整数或浮点数")
    result = 0.0
    for x, y in zip(t1, t2):
        result += x * y
    return float(result)
```

```python
#3.3
def check_valid_mx(my_mx):
    if not isinstance(my_mx, tuple):
        return False
    if len(my_mx) == 0:
        return True
    if not isinstance(my_mx[0], tuple):
        return False
    column_count = len(my_mx[0])
    for row in my_mx:
        if not isinstance(row, tuple):
            return False
        if len(row) != column_count:
            return False
        for value in row:
            if type(value) not in (int, float):
                return False
    return True
```

```python
#3.4
def mx_vec_mult(my_mx, vec):
    if not check_valid_mx(my_mx):
        raise ValueError("输入矩阵不合法")
    if not isinstance(vec, tuple):
        raise TypeError("向量必须是tuple类型")
    for value in vec:
        if type(value) not in (int, float):
            raise TypeError("向量元素必须是整数或浮点数")
    if len(my_mx) == 0:
        return ()
    if len(my_mx[0]) != len(vec):
        raise ValueError("矩阵列数必须等于向量长度")
    result = []
    for row in my_mx:
        value = vec_inner_product(row, vec)
        result.append(value)
    return tuple(result)
```

```python
#4.x
class Vector:
    def __init__(self, dim, elements=None):
        if type(dim) is not int:
            raise TypeError("向量维度必须是整数")
        if dim < 0:
            raise ValueError("向量维度不能为负数")
        if elements is None:
            elements = [0] * dim
        if not isinstance(elements, (list, tuple)):
            raise TypeError("向量元素必须是列表或元组")
        if len(elements) != dim:
            raise ValueError("向量维度与元素数量不一致")
        for value in elements:
            if type(value) not in (int, float):
                raise TypeError("向量元素必须是整数或浮点数")
        self.dim = dim
        self.elements = tuple(elements)

    def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"

    def _check_comparable(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能比较两个Vector对象")
        if self.dim != other.dim:
            raise ValueError("两个向量的维度必须一致")

    def __eq__(self, other):
        self._check_comparable(other)
        return self.elements == other.elements

    def __ne__(self, other):
        self._check_comparable(other)
        return self.elements != other.elements

    def __lt__(self, other):
        self._check_comparable(other)
        return self.elements < other.elements

    def __le__(self, other):
        self._check_comparable(other)
        return self.elements <= other.elements

    def __gt__(self, other):
        self._check_comparable(other)
        return self.elements > other.elements

    def __ge__(self, other):
        self._check_comparable(other)
        return self.elements >= other.elements

    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("dot参数必须是Vector对象")
        if self.dim != other.dim:
            raise ValueError("两个向量的维度必须一致")
        return sum(
            x * y
            for x, y in zip(self.elements, other.elements)
        )

    def __mul__(self, other):
        if type(other) in (int, float):
            new_elements = tuple(
                value * other
                for value in self.elements
            )
            return Vector(self.dim, new_elements)
        if isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("两个向量的维度必须一致")
            new_elements = tuple(
                x * y
                for x, y in zip(
                    self.elements,
                    other.elements
                )
            )
            return Vector(self.dim, new_elements)
        return NotImplemented

    def __rmul__(self, other):
        return self.__mul__(other)
```

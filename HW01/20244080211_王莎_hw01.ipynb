```python
#欧几里得
def gcd(m, n):
    m = abs(m)
    n = abs(n)
    while n != 0:
        m, n = n, m % n
    return m

print("gcd(2018, 2019) =", gcd(2018, 2019))
print("gcd(1200, 300) =", gcd(1200, 300))
print("gcd(5040, 60) =", gcd(5040, 60))
#输入负数时，函数内会转换为正数，因为公约数规定双方为正数
```

```python
#近似欧拉数
import math
def euler_limit(n):
    result = (1 + 1 / n) ** n
    return float(result)

def euler_infinite_sum(n):
    sum_result = 0.0
    factorial = 1
    for k in range(n):
        if k > 0:
            factorial *= k
        sum_result += 1 / factorial
    return float(sum_result)

def euler_approx(epsilon):
    real_e = math.exp(1)
    approximation = 1.0
    k = 0
    term = 1.0
    while abs(real_e - approximation) > epsilon:
        k += 1
        term /= k
        approximation += term
        if term == 0.0:
            break
    return approximation

def print_euler_sum_table(n):
    for k in range(1, n + 1):
        result = euler_infinite_sum(k)
        print(result)
```

```python
#测试整数属性
def is_power_of_2(n):
    while n > 1:
        if n % 2 != 0:
            return False
        n //= 2
    return n == 1
# 测试
for n in [1, 2, 3, 4, 8, 12, 16, 32, 64]:
    print(f"{n}：{is_power_of_2(n)}")

def is_power(b, n):
    if n == 1:
        return True
    if b == 1:
        return False
    while n > 1:
        if n % b != 0:
            return False
        n //= b
    return n == 1

test_cases = [
    (2, 8),
    (2, 12),
    (3, 27),
    (3, 81),
    (5, 125),
    (5, 100),
    (1, 1),
    (1, 5),
    (10, 1)
]
for b, n in test_cases:
    result = is_power(b, n)
    print(f"is_power({b}, {n}) = {result}")
```

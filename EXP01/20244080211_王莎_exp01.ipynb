```python
#2.1
def score_level(score):
    if not 0 <= score <= 100:
        print("成绩需要在0到100之间")
        return "数据错误"

    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"

for score in [95, 85, 70, 50, 101]:
    print(score, score_level(score))
```

```python
#2.2
sum_for = 0
for i in range(1, 101):
    sum_for += i
print("for结果：", sum_for)

sum_while = 0
i = 1
while i <= 100:
    sum_while += i
    i += 1
print("while结果：", sum_while)

print("结果相同：", sum_for == sum_while)
```

```python
#2.3
def factorial_recursive(n):
    if n < 0:
        raise ValueError("阶乘输入不能为负数")
    if n <= 1:
        return 1
    return n * factorial_recursive(n - 1)

def factorial_iterative(n):
    if n < 0:
        raise ValueError("阶乘输入不能为负数")
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

for n in [0, 1, 5, 10]:
    recursive_result = factorial_recursive(n)
    iterative_result = factorial_iterative(n)

    print(
        f"{n}!：递归={recursive_result}, "
        f"迭代={iterative_result}"
    )
```

```python
#2.4
def summary(values):
    if not values:
        raise ValueError("列表不能为空")
    minimum = min(values)
    maximum = max(values)
    average = sum(values) / len(values)
    return minimum, maximum, average

numbers = [10, 20, 30, 40, 50]
minimum, maximum, average = summary(numbers)
print("最小值：", minimum)
print("最大值：", maximum)
print("平均值：", average)
```

```python
#4.2
student_text = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""
student_file = os.path.join(data_dir, "students.txt")
with open(student_file, "w", encoding="utf-8") as file:
    file.write(student_text)
with open(student_file, "r", encoding="utf-8") as file:
    next(file)
    for line in file:
        line = line.strip()
        fields = line.split(",")
        print(fields)
```

```python
#4.3
def parse_student(line):
    fields = line.strip().split(",")
    if len(fields) != 2:
        raise ValueError("字段数量错误，应包含姓名和成绩")
    name = fields[0].strip()
    score_text = fields[1].strip()
    if not name:
        raise ValueError("姓名不能为空")
    try:
        score = int(score_text)
    except ValueError:
        raise ValueError("成绩不是有效整数") from None
    if not 0 <= score <= 100:
        raise ValueError("成绩必须在0到100之间")
    return {
        "name": name,
        "score": score
    }

print(parse_student("Alice,85"))
try:
    print(parse_student("Grace,abc"))
except ValueError as error:
    print("解析错误：", error)
```

```python
#4.4
valid_students = []
invalid_records = []

with open(student_file, "r", encoding="utf-8") as file:
    next(file)
    for line_number, line in enumerate(file, start=2):
        try:
            student = parse_student(line)
            valid_students.append(student)
        except ValueError as error:
            invalid_records.append({
                "line": line_number,
                "reason": str(error)
            })
            print(
                f"第{line_number}行数据无效：{error}"
            )

print("有效记录：", valid_students)
print("无效记录：", invalid_records)
```

```python
#4.5
valid_count = len(valid_students)
invalid_count = len(invalid_records)
valid_scores = [
    student["score"]
    for student in valid_students
]
average_score = sum(valid_scores) / valid_count
passed_count = sum(
    1 for score in valid_scores if score >= 60
)
pass_rate = passed_count / valid_count
report_file = os.path.join(data_dir, "report.txt")
with open(report_file, "w", encoding="utf-8") as file:
    file.write(f"有效记录数：{valid_count}\n")
    file.write(f"无效记录数：{invalid_count}\n")
    file.write(f"平均分：{average_score:.2f}\n")
    file.write(f"及格率：{pass_rate:.2%}\n")

with open(report_file, "a", encoding="utf-8") as file:
    file.write("说明：统计仅包含有效学生记录。\n")
with open(report_file, "r", encoding="utf-8") as file:
    print(file.read())
```

```python
#5.1
class Student:
    def __init__(self, name, score):
        # 初始化学生姓名和成绩
        self.name = name
        self.score = score
    def is_passed(self):
        return self.score >= 60
    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"

student = Student("Alice", 85)
print(student)
print("是否及格：", student.is_passed())
```

```python
#5.2
student_objects = [
    Student(record["name"], record["score"])
    for record in valid_students
]
for student in student_objects:
    print(student)
    print("是否及格：", student.is_passed())
```

```python
#5.3
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic
    def __str__(self):
        return (
            f"GraduateStudent("
            f"name={self.name}, "
            f"score={self.score}, "
            f"research_topic={self.research_topic})"
        )
graduate = GraduateStudent(
    "Alice",
    95,
    "Machine Learning"
)

print(graduate)
print("是否及格：", graduate.is_passed())
```

```python
# 1.1
import re

text = "cat on mat; later: cats and dogs"
digits = re.findall(r"\d+", text)
spaces = re.findall(r"\s+", text)
cats = re.findall(r"\bcat\b", text)

print("原始字符串：", text)
print("数字匹配结果：", digits)
print("空白匹配结果：", spaces)
print("完整单词cat：", cats)
```

```python
# 1.2
import re

text = "cat dog bird CAT"
result = re.findall(r"cat|dog", text)
print("原始字符串：", text)
print("cat|dog匹配结果：", result)
test_text = "aaaa"
result1 = re.findall(r"a|aa|aaa", test_text)
result2 = re.findall(r"aaa|aa|a", test_text)

print("\n测试字符串：", test_text)
print("a|aa|aaa：", result1)
print("aaa|aa|a：", result2)
```

```python
# 1.3 正则表达式量词的使用
import re

text = "ct cat caat caaat"
patterns = [
    r"ca*t",
    r"ca+t",
    r"ca?t",
    r"ca{2}t",
    r"ca{1,2}t"
]
print("原始字符串：", text)
print()

for pattern in patterns:
    result = re.findall(pattern, text)
    print(f"{pattern}：{result}")
```

```python
# 2.1 提取邮箱中的用户名和域名
import re

text = "联系：alice@example.com；备选：bob@test.org"
pattern = r"([A-Za-z0-9._%+-]+)@([A-Za-z0-9.-]+\.[A-Za-z]{2,})"
matches = re.findall(pattern, text)
print("原始字符串：", text)
print()

for username, domain in matches:
    print("用户名：", username)
    print("域名：", domain)
    print("完整邮箱：", username + "@" + domain)
    print("----------------")
```

```python
# 2.2 使用后向引用查找重复单词
import re

text = "cat cat, dog cat, bird bird"
pattern = r"\b(\w+)\s+\1\b"
matches = re.findall(pattern, text)
print("原始字符串：", text)
print("连续重复的单词：", matches)

for match in re.finditer(pattern, text):
    print("完整匹配：", match.group(0))
    print("重复单词：", match.group(1))
```

```python
# 2.3
import re

text = "2024-05-01\n2024/05/02\n2024.05.03"
pattern = r"\b(\d{4})([-/.])(\d{2})\2(\d{2})\b"
replacement = r"\3-\4-\1"
converted = re.sub(pattern, replacement, text)

print("转换前：")
print(text)
print("\n转换后：")
print(converted)
print("\n原始行数：", len(text.splitlines()))
print("转换后行数：", len(converted.splitlines()))
```

```python
# 2.4
import re

text = "联系我：123-456-7890 或 987.654.3210"
pattern = r"\b(\d{3})[-.](\d{3})[-.](\d{4})\b"
replacement = r"(\1)\2-\3"
converted = re.sub(pattern, replacement, text)

print("原始字符串：")
print(text)
print("\n转换后的字符串：")
print(converted)
```

```python
# 3.0

from bs4 import BeautifulSoup

html_doc = """
<html><head><title>实验课程</title></head><body>
<h1>数据分析</h1>
<p class="course">第六讲：正则表达式</p>
<p class="course">第七讲：结构化数据</p>
<a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a>
</body></html>
"""

print("原始HTML：")
print(html_doc)
```

```python
# 3.1 构建BeautifulSoup解析树
from bs4 import BeautifulSoup

soup = BeautifulSoup(html_doc, "html.parser")
title_tag = soup.find("title")
page_title = title_tag.get_text()
h1_tag = soup.find("h1")
h1_name = h1_tag.name
h1_text = h1_tag.get_text()

print("页面标题：", page_title)
print("h1标签名称：", h1_name)
print("h1标签文本：", h1_text)
```

```python
# 3.2 查找课程段落与超链接

course_paragraphs = soup.find_all(
    "p",
    class_="course"
)
print("课程段落：")
for index, paragraph in enumerate(
    course_paragraphs, start=1
):
    print(f"第{index}个段落：{paragraph.get_text()}")
links = soup.find_all("a")
print("\n页面链接：")
for index, link in enumerate(links, start=1):
    link_text = link.get_text()
    link_url = link.get("href")

    print(f"第{index}个链接：")
    print("链接文本：", link_text)
    print("链接地址：", link_url)
    print("----------------")
```

```python
# 3.3 使用get_text()提取HTML文本
body_text = soup.body.get_text(
    separator="\n",
    strip=True
)
document_text = soup.get_text(
    separator="\n",
    strip=True
)
paragraph_text = soup.find(
    "p",
    class_="course"
).get_text(strip=True)
link_text = soup.find("a").get_text(strip=True)
print("整个HTML文档的文本：")
print(document_text)
print("\nbody中的可见文本：")
print(body_text)
print("\n单个段落的文本：")
print(paragraph_text)
print("\n单个链接的文本：")
print(link_text)
```
